 Data Cleaning

```python
df['Company_Name'].fillna('Not Assigned', inplace=True)
df['Position'].fillna('Not Assigned', inplace=True)
df['City'].fillna('Unknown', inplace=True)
```

# Exploratory Data Analysis (EDA)

```python
# Placement Status
df['Placement_Status'].value_counts()
```

```python
# Average Package by course
df.groupby('Course_Name')['Package_LPA'].mean()
```

# Data Visualization

1) Placement Status Analysis 

```python
df['Placement_Status'].value_counts().plot(kind='bar')
plt.title("Placement Status Distribution")
plt.xlabel("Placement Status")
plt.ylabel("Number of Students")
plt.show()
```

2) Course- Wise Student Count (Bar Chart)

```python
df['Course_Name'].value_counts().plot(kind='bar')
plt.title("Number of Students per Course")
plt.xlabel("Course Name")
plt.ylabel("Number of Students")
plt.show()
```

3) Average Package by Course 

```python
df.groupby('Course_Name')['Package_LPA'].mean().plot(kind='line')
plt.title("Average Package by Course")
plt.xlabel("Course Name")
plt.ylabel("Average Package (LPA)")
plt.show()
```

4) Degree Marks Distribution (Histogram)

```python

plt.hist(df['Degree_Marks'], bins=10)  # Changed scatter to hist which only needs one array
plt.title("Distribution of Degree Marks")
plt.xlabel("Degree Marks")
plt.ylabel("Number of Students")
plt.show()

```

5) Package Distribution 

```python
plt.hist(df['Package_LPA'], bins=10)
plt.title("Package Distribution")
plt.xlabel("Package (LPA)")
plt.ylabel("Number of Students")
plt.show()
```

6) Degree Marks vs Package 

```python
plt.scatter(df['Degree_Marks'], df['Package_LPA'])
plt.title("Degree Marks vs Package")
plt.xlabel("Degree Marks")
plt.ylabel("Package (LPA)")
plt.show()
```

7) City - wise job Distribution

```python
df['City'].value_counts().plot(kind='bar')
plt.title("City-wise Job Distribution")
plt.xlabel("City")
plt.ylabel("Number of Students")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 18))

# 1️⃣ Placement Status
plt.subplot(3, 2, 1)
df['Placement_Status'].value_counts().plot(kind='bar')
plt.title("Placement Status Distribution")
plt.xlabel("Status")
plt.ylabel("Students")

# 2️⃣ Students per Course
plt.subplot(3, 2, 2)
df['Course_Name'].value_counts().plot(kind='bar')
plt.title("Students per Course")
plt.xlabel("Course")
plt.ylabel("Students")

# 3️⃣ Average Package by Course
plt.subplot(3, 2, 3)
df.groupby('Course_Name')['Package_LPA'].mean().plot(kind='bar')
plt.title("Average Package by Course")
plt.xlabel("Course")
plt.ylabel("Package (LPA)")

# 4️⃣ Degree Marks Distribution
plt.subplot(3, 2, 4)
plt.hist(df['Degree_Marks'], bins=10)
plt.title("Degree Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Students")

# 5️⃣ Package Distribution
plt.subplot(3, 2, 5)
plt.hist(df['Package_LPA'], bins=10)
plt.title("Package Distribution")
plt.xlabel("Package (LPA)")
plt.ylabel("Students")

# 6️⃣ Marks vs Package
plt.subplot(3, 2, 6)
plt.scatter(df['Degree_Marks'], df['Package_LPA'])
plt.title("Marks vs Package")
plt.xlabel("Marks")
plt.ylabel("Package (LPA)")

plt.tight_layout()
plt.show()
```
